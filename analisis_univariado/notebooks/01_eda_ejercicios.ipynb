## 6. Medidas de dispersión

**¿los valores están muy juntos o muy separados?**


### Actividad 8
Calcula para `fare`:

Usar `.min()`, `.max()`, `.var()`, `.std()` y `np.ptp(...)`.

## 9. Índice de Gini de heterogeneidad

Este índice se usa con variables categóricas para medir qué tan repartidas están las categorías.



### Actividad 16
Exploa la columna `class`:

1. Realiza el cálculo del índice Gini para la variable class




```python
gini_index = 1 - (frecuencia_relativa ** 2).sum()

print(f"El Índice de Gini para 'class' es: {gini_index}")
```

**Respuestas:**

## 12. Opcional

Utiliza la variable  `survived` u otra que sea de tu interés para calcular:
- tabla de frecuencias

- índice de Gini de heterogeneidad

```python
frecuencias  = df['survived'].value_counts()
print(f"La frecuencia es: {frecuencias}")

proporciones = df['survived'].value_counts(normalize=True)
print(f"La proporciones es: {proporciones}")

gini_index_e = 1 - (proporciones ** 2).sum()

print(f"El Índice de Gini {gini_index_e}")
```

Con la frecuencia se observa que la mayoría de los pasajeros no sobrevivieron, con el índice que Gini,vemos que los datos en las categorías no es uniforme.
